```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
df = pd.read_csv("data/IRIS.csv")

df.head()
```

```python
df.shape

df.info()

df.describe()

df.isnull().sum()

df.columns
```

```python
#--- distribution of species ---
df["species"].value_counts()
```

```python
# --visualization--
# --- distribution of sepal length ---
plt.figure(figsize=(8,5))

sns.histplot(
    df["sepal_length"],
    bins=20,
    kde=True
)

plt.title("Sepal Length Distribution")

plt.tight_layout()

plt.savefig("visuals/sepal_length_distribution.png")

plt.show()
```

```python
#--- species comparison ---
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="sepal_length",
    y="petal_length",
    hue="species"
)

plt.title("Species Comparison")

plt.tight_layout()

plt.savefig("visuals/species_comparison.png")

plt.show()
```

```python
#--- correlation heatmap ---
plt.figure(figsize=(8,6))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.savefig("visuals/correlation_heatmap.png")

plt.show()
```

```python
#--- boxplot of petal length by species ---
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x="species",
    y="petal_length"
)

plt.title("Petal Length by Species")

plt.tight_layout()

plt.savefig("visuals/boxplot_species.png")

plt.show()
```

```python
#--- pairplot of all features colored by species ---
sns.pairplot(
    df,
    hue="species"
)

plt.savefig("visuals/species_pairplot.png")

plt.show()
```

```python
#--- group by species and calculate mean of each feature ---
df.groupby("species").mean()
```

# Key Insights

- Iris-setosa is clearly separable from other species.
- Petal dimensions show strong correlation.
- Iris-versicolor and Iris-virginica have overlapping characteristics.
- Petal length is one of the strongest distinguishing features.
