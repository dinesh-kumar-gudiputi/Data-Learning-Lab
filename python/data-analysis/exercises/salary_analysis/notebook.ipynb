```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
df = pd.read_csv("data/Salary_Data.csv")

df.head()
```

```python
df.shape

df.info()

df.describe()

df.isnull().sum()

df.columns
```

```python
#--visualization--
#--- distribution of salary ---
plt.figure(figsize=(8,5))

sns.histplot(
    df["Salary"],
    bins=20,
    kde=True
)

plt.title("Salary Distribution")

plt.xlabel("Salary")
plt.ylabel("Count")

plt.tight_layout()

plt.savefig("visuals/salary_distribution.png")

plt.show()
```

```python
#--- experience vs salary ---
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="YearsExperience",
    y="Salary"
)

plt.title("Experience vs Salary")

plt.xlabel("Years of Experience")
plt.ylabel("Salary")

plt.tight_layout()

plt.savefig("visuals/experience_vs_salary.png")

plt.show()
```

```python
#--- salary growth with experience ---
sorted_df = df.sort_values("YearsExperience")

plt.figure(figsize=(10,5))

plt.plot(
    sorted_df["YearsExperience"],
    sorted_df["Salary"],
    marker="o"
)

plt.title("Salary Growth with Experience")

plt.xlabel("Years of Experience")
plt.ylabel("Salary")

plt.tight_layout()

plt.savefig("visuals/salary_growth.png")

plt.show()
```

```python
#--- correlation heatmap ---
df.corr(numeric_only=True)
```

```python
#--- correlation heatmap ---
plt.figure(figsize=(6,4))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.savefig("visuals/correlation_heatmap.png")

plt.show()
```

```python
#--- top salary records ---
top_salary = df.sort_values(
    by="Salary",
    ascending=False
).head(10)

plt.figure(figsize=(10,5))

sns.barplot(
    x=top_salary["YearsExperience"],
    y=top_salary["Salary"]
)

plt.title("Top Salary Records")

plt.xlabel("Years of Experience")
plt.ylabel("Salary")

plt.tight_layout()

plt.savefig("visuals/top_salary_roles.png")

plt.show()
```
